```python

df = df.rename(columns={"PAY_0": "PAY_1", "default payment next month": "default"})

pay_cols     = ["PAY_6", "PAY_5", "PAY_4", "PAY_3", "PAY_2", "PAY_1"]
bill_cols    = [f"BILL_AMT{i}" for i in range(6, 0, -1)]
payamt_cols  = [f"PAY_AMT{i}" for i in range(6, 0, -1)]

print(pay_cols)
print(bill_cols)
print(payamt_cols)
```

```python
import matplotlib.pyplot as plt

months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
i = 0  

fig, ax = plt.subplots(1, 3, figsize=(14, 3.5))
ax[0].plot(months, df.loc[i, pay_cols].values, marker="o")
ax[0].set_title("PAY (delay status)")
ax[1].plot(months, df.loc[i, bill_cols].values, marker="o")
ax[1].set_title("BILL_AMT (bill)")
ax[2].plot(months, df.loc[i, payamt_cols].values, marker="o")
ax[2].set_title("PAY_AMT (payment)")
plt.tight_layout()
plt.show()

print("default =", df.loc[i, "default"])
```

```python
def plot_customer(i):
    fig, ax = plt.subplots(1, 3, figsize=(14, 3))
    ax[0].plot(months, df.loc[i, pay_cols].values, marker="o")
    ax[0].set_title(f"Customer {i}: PAY")
    ax[1].plot(months, df.loc[i, bill_cols].values, marker="o")
    ax[1].set_title("BILL_AMT")
    ax[2].plot(months, df.loc[i, payamt_cols].values, marker="o")
    ax[2].set_title("PAY_AMT")
    plt.tight_layout()
    plt.show()
    print("default =", df.loc[i, "default"])

for i in df[df["default"] == 1].index[:2]:
    plot_customer(i)

for i in df[df["default"] == 0].index[:2]:
    plot_customer(i)
```

```python
print(df.groupby("SEX")["default"].mean().round(3))
print()
print(df.groupby("EDUCATION")["default"].mean().round(3))
print()
df["age_group"] = pd.cut(df["AGE"], bins=[20, 30, 40, 50, 60, 80])
print(df.groupby("age_group", observed=True)["default"].agg(["count", "mean"]).round(3))
```

```python
# 1. Ուշացումների միջինը ըստ դեֆոլտի
print("1) Միջին PAY ըստ default-ի")
print(df.groupby("default")[pay_cols].mean().round(2))
print()

# 2. Քանի ամսում է ուշացրել
df["n_delayed"] = (df[pay_cols] > 0).sum(axis=1)
print("2) Դեֆոլտի բաժինը ըստ ուշացած ամիսների քանակի")
print(df.groupby("n_delayed")["default"].agg(["count", "mean"]).round(3))
print()

# 3. Ստատիկ պարամետրեր
print("3) Ըստ սեռի")
print(df.groupby("SEX")["default"].mean().round(3))
print()
print("4) Ըստ կրթության")
print(df.groupby("EDUCATION")["default"].mean().round(3))
print()
df["age_group"] = pd.cut(df["AGE"], bins=[20, 30, 40, 50, 60, 80])
print("5) Ըստ տարիքային խմբի")
print(df.groupby("age_group", observed=True)["default"].agg(["count", "mean"]).round(3))
```
